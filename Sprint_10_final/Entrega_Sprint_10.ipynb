## Crear una función que dada la fecha calcule la edad a día de hoy

```python
from datetime import datetime

def calcular_edad(fecha_nacimiento):
    hoy = datetime.today()
    
    edad = hoy.year - fecha_nacimiento.year
    
    # Ajuste si aún no ha cumplido años este año
    if (hoy.month, hoy.day) < (fecha_nacimiento.month, fecha_nacimiento.day):
        edad -= 1
        
    return edad
```

## Crea una columna nueva en el dataframe con la edad

```python
df["Edad"] = df["Fecha_de_nacimiento"].apply(calcular_edad)

df.head(5)
```

## Creación de una función para generar gráficos automáticamente

Se crea una función que recibe un DataFrame como parámetro y genera un gráfico para cada columna, según su tipo de dato:

- **Variables numéricas** → histograma o boxplot  
- **Variables categóricas** → gráfico de barras con los valores más frecuentes  
- **Variables de tipo fecha** → gráfico de barras con los años más frecuentes  

```python
import matplotlib.pyplot as plt
import pandas as pd

def graficos_basicos(df):
    
    for col in df.columns:
        
        print("Gráfico de:", col)
        
        plt.figure() # Esto me crear un gráfico nuevo cada vez, sino los mezclaría
        
        # Si es número
        if df[col].dtype == "int64" or df[col].dtype == "float64":
            df[col].hist()
        
        # Si es fecha
        elif "datetime" in str(df[col].dtype):
            df[col].dt.year.value_counts().plot(kind="bar") # dt.year me da el año
                                                            # value_counts cuenta cuántos por año
        # Si es texto (categoría)
        else:
            df[col].value_counts().head(10).plot(kind="bar")    # coge los 10 más frecuentes
        
        plt.title(col)  # Esto pone el nombre de la columna como título de gráfico
        plt.show()
```

## Construcción de una función para calcular una ruta mínima

La función debe:

- Recibir la matriz de distancias  
- Recibir una ciudad de origen  
- Generar una propuesta de ruta lo más corta posible  
- Devolver:
    - el orden de visita de las ciudades
    - la distancia total recorrida

```python
def ruta(df, origen):

    # Variables que necesito

    visitadas = [origen]
    
    ciudad_actual = origen
    
    distancia_total = 0

    # Hasta que no visitamos todas la ciudades 

    while len(visitadas) < len(df):

        distancias = df.loc[ciudad_actual]  # Nos da la distancia desde la ciudad actual

        distancias = distancias.drop(visitadas) # Elimina las ciudades visitadas

        siguiente = distancias.idxmin() # Nos da la siguiente ciudad más cercana, índice del valor mínimo

        distancia_total += distancias[siguiente]    # Actualiza la distancia

        visitadas.append(siguiente) # Añade la ciudad

        ciudad_actual = siguiente   # Pasa a la sigiente ciudad

    return visitadas, distancia_total
```

```python
ruta_final, distancia = ruta(df, "Barcelona")

print("Ruta:\n")

for ciudad in ruta_final:
    print(ciudad)

print("\nDistancia total:", distancia)
```

```python
ruta_final, distancia = ruta(df, "Valencia")

print("Ruta:\n")

for ciudad in ruta_final:
    print(ciudad)

print("\nDistancia total:", distancia)
```

## Búsqueda de la mejor ciudad inicial

Se prueba el algoritmo comenzando desde cada ciudad disponible en la matriz de distancias.

Para cada ciudad:

- Se calcula la ruta propuesta
- Se obtiene la distancia total recorrida
- Se compara con la mejor distancia encontrada hasta el momento

Finalmente, se conserva la ciudad inicial que genera la ruta más corta según el algoritmo implementado.

```python
# Variables para guardar la mejor ruta

mejor_distancia = float("inf")  # Lo usamos porque cualquier distancia real será menor.

mejor_ciudad = None

mejor_ruta = None


# Recorremos todas la ciudades

for ciudad in df.index: # df.index contiene todas las ciudades

    ruta_final, distancia = ruta(df, ciudad)    # prueba la ruta empezando desde esa ciudad.

    if distancia < mejor_distancia: # comparamos las distancias y guardamos la mejor ruta

        mejor_distancia = distancia

        mejor_ciudad = ciudad

        mejor_ruta = ruta_final
```

```python
print("Mejor ciudad inicial:", mejor_ciudad)

print("\nDistancia total:", mejor_distancia)

print("\nRuta:\n")

for ciudad in mejor_ruta:
    print(ciudad)
```
