# Variable Numérica

```python
import matplotlib.pyplot as plt


df_transactions["amount"].plot.hist(bins=30)

# Aprovecho matplotlib para poner títulos a los gráficos

plt.title("Distribución de amount")
plt.xlabel("Cantidad")

plt.show()
```

## Interpretación del gráfico:

La mayoría de las transacciones se concentran aproximadamente entre 100 y 400 unidades de cantidad (amount).

La distribución presenta una ligera asimetría hacia la derecha, ya que existen algunas transacciones con importes más elevados, aunque son menos frecuentes.

Los valores muy altos (por encima de 600) aparecen de forma aislada, por lo que podrían considerarse valores atípicos o poco comunes.

# Variable Categórica

```python
df_transactions["declined"].value_counts().plot.bar()

plt.title("Transacciones rechazadas")

plt.show()
```

## Interpretación del gráfico:

La gran mayoría de las transacciones fueron aceptadas (declined = 0), mientras que las transacciones rechazadas (declined = 1) representan una proporción muy pequeña del total.

Esto indica que el sistema de pagos tiene una tasa de rechazo muy baja.

# Dos variables numéricas

```python
sample = df_transactions.sample(1000)

# Se toma una muestra de 1000 debido al tamaño del dataset

sample.plot(kind="scatter", x="lat", y="amount", alpha=0.3)

plt.title("Latitud vs Amount")

plt.show()
```

## Interpretación del gráfico:

No se observa una relación clara entre la latitud y el importe de las transacciones.

Los puntos aparecen dispersos de forma aleatoria y no siguen una tendencia ascendente ni descendente.

Esto indica que la localización geográfica (latitud) no parece influir directamente en el importe de las compras.

# Dos variables categóricas

```python
# TOP 5 países
top_paises = df_users["country"].value_counts().head(5).index

# Filtrar países
df_filtrado = df_users[df_users["country"].isin(top_paises)]

# TOP 5 ciudades dentro de esos países
top_ciudades = df_filtrado["city"].value_counts().head(5).index

# Filtrar ciudades
df_filtrado = df_filtrado[df_filtrado["city"].isin(top_ciudades)]

# Crosstab
tabla = pd.crosstab(
    df_filtrado["country"],
    df_filtrado["city"]
)

# Gráfico
tabla.plot.bar(stacked=True)

plt.title("Top ciudades en los principales países")

plt.xticks(rotation=45) # inclinamos las etiquetas 45

plt.show()
```

## Interpretación del gráfico:

El gráfico muestra la distribución de las ciudades más frecuentes dentro de los principales países analizados.

En Países Bajos predominan las ciudades de Groningen y Utrecht, mientras que en España destacan Madrid y Sevilla.

En Reino Unido, Manchester concentra la mayor parte de los usuarios representados en el gráfico.

Esto indica que algunos países presentan una mayor concentración de usuarios en determinadas ciudades principales.

# Una variable categórica y una numérica

```python
df_merge = pd.merge(
    df_transactions,
    df_users,
    left_on="user_id",
    right_on="id"
)
print(df_merge.head())
```

```python
media_pais = df_merge.groupby("country")["amount"].mean().round(2)

print(media_pais)
```

```python
media_pais.sort_values().plot.bar(title="Importe medio por país")
```

## Interpretación del gráfico

El gráfico muestra que el importe medio de las transacciones es similar en todos los países analizados. Estados Unidos presenta el valor medio ligeramente más alto y Francia el más bajo, aunque las diferencias son pequeñas, lo que indica un comportamiento de gasto relativamente uniforme entre países.

# Combinación de 3 variables

```python
sample = df_transactions.sample(1000)

# Transacciones aceptadas
aceptadas = sample[sample["declined"] == 0]

# Transacciones rechazadas
rechazadas = sample[sample["declined"] == 1]

# Scatter aceptadas
plt.scatter(
    aceptadas["lat"],
    aceptadas["amount"],
    alpha=0.5,
    label="Aceptadas"
)

# Scatter rechazadas
plt.scatter(
    rechazadas["lat"],
    rechazadas["amount"],
    alpha=0.8,
    label="Rechazadas"
)

plt.title("Latitud, amount y declined")

plt.xlabel("Latitud")
plt.ylabel("Amount")

plt.legend()

plt.show()
```

## Interpretación del gráfico

El gráfico representa la relación entre la latitud y el importe de las transacciones, diferenciando las operaciones aceptadas y rechazadas.

La mayoría de las transacciones corresponden a operaciones aceptadas, mientras que las rechazadas representan una proporción muy pequeña del total.

No se observa una relación clara entre la ubicación geográfica (latitud) y el importe de las transacciones, ya que los puntos aparecen distribuidos de forma dispersa.

Las pocas transacciones rechazadas también aparecen repartidas sin seguir un patrón específico, lo que sugiere que el rechazo no depende directamente de la latitud ni del importe.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Seleccionar variables numéricas
corr = df_transactions[
    ['amount', 'lat', 'longitude', 'declined']
].corr()

# Heatmap
sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlación entre variables')

plt.show()
```

## Interpretación del gráfico

La matriz de correlación muestra que no existen relaciones lineales fuertes entre la mayoría de variables analizadas. 

La única correlación moderada aparece entre latitud y longitud (0.5), debido a la agrupación geográfica de las transacciones. 

El importe (amount) y el estado de rechazo (declined) presentan correlaciones muy próximas a cero con el resto de variables, lo que indica ausencia de relación significativa.
