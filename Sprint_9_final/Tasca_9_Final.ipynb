# Cálculo del Índice de Masa Corporal (IMC)

## Introducción

En este ejercicio se desarrolla un programa en Python para calcular el Índice de Masa Corporal (IMC) de un usuario.

A partir del peso y la altura introducidos, el programa:

- Calcula el IMC
- Clasifica el resultado según categorías estándar
- Muestra el resultado por pantalla

El programa está estructurado en funciones para separar la entrada de datos, el cálculo, la clasificación y la salida.

```python
# Imports
```

## Entrada de datos

Esta función solicita al usuario su peso y altura.
En caso de error en la introducción de datos, devuelve valores nulos.

```python
def pedir_datos():
    """
    Solicita al usuario el peso y la altura.
    Devuelve ambos valores como float o None si hay error.
    """
    try:
        peso = float(input("Introduce tu peso (kg): "))
        altura = float(input("Introduce tu altura (m): "))
        return peso, altura
    except ValueError:
        return None, None
```

## Cálculo del IMC

Esta función calcula el Índice de Masa Corporal a partir del peso y la altura.

Si los valores no son válidos, devuelve None.

```python
def calcular_imc(peso, altura):
    """
    Calcula el IMC usando la fórmula: peso / altura^2.
    """
    if peso is None or altura is None:
        return None
    if peso <= 0 or altura <= 0:
        return None
    return peso / (altura ** 2)
```

## Clasificación del IMC

Esta función clasifica el IMC en diferentes categorías según los valores establecidos.

```python
def clasificar_imc(imc):
    """
    Clasifica el IMC en categorías estándar.
    """
    if imc is None:
        return None

    if imc < 18.5:
        return "Bajo peso"
    elif imc <= 24.9:
        return "Peso normal"
    elif imc <= 29.9:
        return "Sobrepeso"
    else:
        return "Obesidad"
```

## Visualización de resultados

Esta función muestra el IMC calculado y su categoría correspondiente.

```python
def mostrar_resultado(imc, categoria):
    """
    Muestra el resultado del IMC y su clasificación.
    """
    if imc is None or categoria is None:
        print("Datos no válidos")
    else:
        print("Tu IMC es:", round(imc, 2))
        print("Tu categoría es:", categoria)
```

## Función principal

Esta función coordina todo el flujo del programa:

1. Solicita los datos
2. Calcula el IMC
3. Clasifica el resultado
4. Muestra la información

```python
def main():
    """
    Función principal que coordina el programa.
    """
    peso, altura = pedir_datos()
    imc = calcular_imc(peso, altura)
    categoria = clasificar_imc(imc)

    # Redondear solo si el IMC es válido
    if imc is not None:
        imc_redondeado = round(imc, 2)
    else:
        imc_redondeado = None

    resultado = (imc_redondeado, categoria)

    mostrar_resultado(imc, categoria)

    return resultado
```

## Ejecución del programa

```python
resultado_final = main()
print("Resultado guardado:", resultado_final)
```

## Conclusiones

Este ejercicio muestra cómo estructurar un programa en Python utilizando funciones con responsabilidades específicas.

La separación entre entrada, procesamiento y salida permite un código más claro, reutilizable y fácil de mantener.

Además, se han aplicado validaciones básicas para evitar errores en los datos introducidos por el usuario.

# Convertidor de temperatura

## Introducción

En este ejercicio se desarrolla un programa en Python que permite convertir una temperatura entre distintas unidades:

- Celsius (C)
- Fahrenheit (F)
- Kelvin (K)

El programa solicita al usuario una temperatura y su unidad, normaliza la entrada y calcula las equivalencias en las otras unidades.

Se ha estructurado en funciones para separar la entrada de datos, la normalización, el cálculo y la visualización.

```python
# Imports
```

## Entrada de datos

Esta función solicita al usuario la temperatura y la unidad.
Si se introduce un valor incorrecto, devuelve valores nulos.

```python
def pedir_datos():
    """
    Solicita la temperatura y la unidad al usuario.
    Devuelve los valores o None si hay error.
    """
    try:
        temp = float(input("Introduce la temperatura: "))
        unidad = input("Unidad (C, F, K): ").upper()
        return temp, unidad
    except ValueError:
        return None, None
```

## Normalización de la unidad

Esta función convierte la unidad introducida en un formato estándar (C, F o K),
permitiendo distintas formas de entrada como "celsius" o "C".

```python
def normalizar_unidad(unidad):
    """
    Normaliza la unidad de temperatura a C, F o K.
    """
    unidad = unidad.upper()

    if unidad in ["C", "CELSIUS"]:
        return "C"
    elif unidad in ["F", "FAHRENHEIT"]:
        return "F"
    elif unidad in ["K", "KELVIN"]:
        return "K"
    else:
        return None
```

## Conversión de temperatura

Esta función convierte la temperatura a Celsius y, a partir de ahí,
calcula su equivalente en Fahrenheit y Kelvin.

```python
def convertir(temp, unidad):
    """
    Convierte una temperatura a Celsius, Fahrenheit y Kelvin.
    """
    unidad = normalizar_unidad(unidad)

    if unidad is None or temp is None:
        return None

    if unidad == "C":
        c = temp
    elif unidad == "F":
        c = (temp - 32) * 5/9
    elif unidad == "K":
        c = temp - 273.15

    return {
        "Celsius": round(c, 2),
        "Fahrenheit": round(c * 9/5 + 32, 2),
        "Kelvin": round(c + 273.15, 2)
    }
```

## Visualización de resultados

Esta función muestra el resultado de la conversión.

```python
def mostrar_resultado(resultado):
    """
    Muestra el resultado de la conversión.
    """
    if resultado is None:
        print("Unidad no válida o datos incorrectos")
    else:
        print("Resultado =", resultado)
```

## Función principal

Esta función coordina el flujo del programa:

1. Solicita los datos
2. Convierte la temperatura
3. Muestra el resultado

```python
def main():
    """
    Función principal del programa.
    """
    temp, unidad = pedir_datos()
    resultado = convertir(temp, unidad)

    mostrar_resultado(resultado)

    return resultado
```

## Ejecución del programa

```python
resultado_final = main()
```

## Conclusiones

Este ejercicio muestra cómo trabajar con diferentes unidades y transformar datos mediante funciones.

La normalización de la entrada permite aceptar distintas formas de escribir las unidades, mejorando el programa.

La estructura modular facilita la comprensión, reutilización y mantenimiento del código.

# Contador de palabras de un texto

## Introducción

En este ejercicio se desarrolla un programa en Python que permite contar cuántas veces aparece cada palabra en un texto introducido por el usuario.

El programa realiza los siguientes pasos:

- Convierte el texto a minúsculas
- Elimina signos de puntuación
- Divide el texto en palabras
- Cuenta la frecuencia de cada palabra utilizando un diccionario


```python
# Imports
```

## Procesamiento del texto

Esta función recibe un texto y realiza:

- Normalización a minúsculas
- Eliminación de signos de puntuación
- Separación en palabras
- Conteo de frecuencia de cada palabra

```python
def contador_palabras(texto):
    """
    Cuenta cuántas veces aparece cada palabra en un texto.
    """

    # pasar a minúsculas
    texto = texto.lower()

    # limpiar puntuación
    texto = texto.replace(",", "")
    texto = texto.replace(".", "")
    texto = texto.replace(";", "")
    texto = texto.replace(":", "")

    # separar las palabras
    palabras = texto.split()

    # diccionario contador
    palabras_contadas = {}

    for palabra in palabras:
        if palabra in palabras_contadas:
            palabras_contadas[palabra] += 1
        else:
            palabras_contadas[palabra] = 1

    return palabras_contadas
```

## Función principal

Esta función solicita un texto al usuario, aplica el contador de palabras y muestra el resultado.

```python
def main():
    """
    Función principal del programa.
    """
    texto = input("Introduce tu texto: ")

    resultado = contador_palabras(texto)

    print("Conteo de palabras:", resultado)
```

## Ejecución del programa

```python
main()
```

Este ejercicio muestra cómo procesar texto en Python de forma sencilla utilizando operaciones básicas con cadenas y diccionarios.

# Longitud media de las palabras de un texto

## Introducción

En este ejercicio se desarrolla un programa en Python que calcula la longitud media de las palabras de un texto introducido por el usuario.

El proceso incluye:

- Normalización del texto a minúsculas
- Eliminación de signos de puntuación
- Separación en palabras
- Cálculo del número total de letras
- Obtención de la media de caracteres por palabra


```python
# Imports
```

## Procesamiento del texto

Esta función recibe un texto y calcula la longitud media de sus palabras.

Para ello:

- Normaliza el texto a minúsculas
- Elimina signos de puntuación
- Divide el texto en palabras
- Suma la longitud de cada palabra
- Calcula la media

```python
def longitud_media_palabras(texto):
    """
    Calcula la longitud media de las palabras de un texto.
    """

    # normalizar texto
    texto = texto.lower()

    # limpiar puntuación
    for signo in [",", ".", ";", ":"]:
        texto = texto.replace(signo, "")

    # separar las palabras
    palabras = texto.split()

    # comprobar si no hay palabras
    if len(palabras) == 0:
        return 0

    # crear contador
    total_letras = 0

    for palabra in palabras:
        letras = len(palabra)
        total_letras = total_letras + letras

    # calcular media
    media = total_letras / len(palabras)

    return round(media, 2)
```

## Función principal

Esta función solicita un texto al usuario, calcula la longitud media de sus palabras y muestra el resultado.

```python
def main():
    """
    Función principal del programa.
    """
    texto = input("Introduce tu texto: ")

    resultado = longitud_media_palabras(texto)

    print("Longitud media:", resultado)
```

## Ejecución del programa

```python
main()
```

## Conclusiones

Este ejercicio permite trabajar con cadenas de texto y aplicar operaciones básicas de procesamiento.

El uso de bucles y funciones facilita el cálculo de métricas como la longitud media, que son comunes en análisis de datos textuales.

# Inversión de un diccionario y detección de duplicados

## Introducción

En este ejercicio se desarrolla un programa en Python que permite invertir un diccionario.

El objetivo es:

- Intercambiar claves por valores
- Agrupar en listas las claves que tienen el mismo valor
- Detectar valores duplicados (aquellos que aparecen más de una vez)

Este tipo de operación es útil en análisis de datos cuando se quiere reorganizar la información.

```python
# Imports
```

## Inversión del diccionario

Esta función recorre un diccionario original y crea uno nuevo donde:

- Las claves pasan a ser valores
- Los valores pasan a ser claves
- Si un valor se repite, se agrupan las claves en una lista

Además, se identifican los valores duplicados.

```python
def invertir_diccionario(dic):
    """
    Invierte un diccionario agrupando claves con el mismo valor
    y detecta valores duplicados.
    """

    inv_dic = {}
    duplicados = {}

    for clave, valor in dic.items():

        if valor not in inv_dic:
            inv_dic[valor] = [clave]
        else:
            inv_dic[valor].append(clave)

            if valor not in duplicados:
                duplicados[valor] = inv_dic[valor][:]  # copia de la lista

    return inv_dic, duplicados
```

## Función principal

Esta función permite al usuario introducir un diccionario manualmente.

Para ello:

1. Se solicita el número de elementos
2. Se introducen las claves y valores
3. Se invierte el diccionario
4. Se muestran los resultados

```python
def main():
    """
    Función principal del programa.
    """
    dic = {}

    n = int(input("¿Cuántos elementos quieres introducir? "))

    for i in range(n):
        clave = input("Introduce clave: ")
        valor = input("Introduce valor: ")
        dic[clave] = valor

    inv_dic, duplicados = invertir_diccionario(dic)

    print("Diccionario invertido:", inv_dic)
    print("Valores duplicados:", duplicados)
```

## Ejecución del programa

```python
main()
```

## Ejemplo de prueba

A continuación se muestra un ejemplo de diccionario para probar la función sin introducir datos manualmente.

```python
dic = {
    "a": 1,
    "b": 2,
    "c": 1,
    "d": 3,
    "e": 2
}

inv_dic, duplicados = invertir_diccionario(dic)

print("Diccionario invertido:", inv_dic)
print("Valores duplicados:", duplicados)
```

## Conclusiones

Este ejercicio permite trabajar con diccionarios y comprender cómo reorganizar datos.

El uso de listas dentro de diccionarios permite agrupar información de forma eficiente.

Además, la detección de duplicados es útil para identificar patrones o valores repetidos en conjuntos de datos.

# Análisis de palabras de un texto

## Introducción

En este ejercicio se desarrolla un programa en Python que analiza un texto contenido en un archivo.

El programa realiza las siguientes operaciones:

- Lee el contenido del archivo
- Limpia el texto (minúsculas y eliminación de signos)
- Cuenta la frecuencia de cada palabra
- Agrupa las palabras por su letra inicial
- Ordena los resultados alfabéticamente

Este tipo de procesamiento es habitual en análisis de texto y tratamiento de datos.

```python
# Imports

from pathlib import Path
```

## Lectura del archivo

Esta función abre el archivo de texto y devuelve su contenido completo.

```python
def leer_archivo():
    ruta = Path.cwd().parent / "tu me quieres blanca.txt"
    
    if not ruta.exists():
        print("Ruta usada:", ruta)
        raise FileNotFoundError("Archivo no encontrado")
    
    return ruta.read_text(encoding="utf-8")
```

## Limpieza del texto

Esta función normaliza el texto:

- Convierte todo a minúsculas
- Elimina signos de puntuación

```python
def limpiar_texto(texto):
    """
    Limpia el texto eliminando signos y pasando a minúsculas.
    """
    texto = texto.lower()

    signos = ".,;:¡!¿?()\""
    for signo in signos:
        texto = texto.replace(signo, "")

    return texto
```

## Conteo de palabras

Esta función divide el texto en palabras y cuenta cuántas veces aparece cada una.

```python
def contar_palabras(texto):
    """
    Cuenta la frecuencia de cada palabra en el texto.
    """
    palabras = texto.split()
    conteo = {}

    for palabra in palabras:
        if palabra in conteo:
            conteo[palabra] += 1
        else:
            conteo[palabra] = 1

    return conteo
```

## Agrupación por letra inicial

Esta función agrupa las palabras según su primera letra.

```python
def agrupar_por_letra(conteo):
    """
    Agrupa palabras según su letra inicial.
    """
    resultado = {}

    for palabra, cantidad in conteo.items():
        letra = palabra[0]

        if letra not in resultado:
            resultado[letra] = {}

        resultado[letra][palabra] = cantidad

    return resultado
```

## Ordenación de resultados

Esta función ordena:

- Las letras del diccionario
- Las palabras dentro de cada grupo

```python
def ordenar_resultado(diccionario):
    """
    Ordena el diccionario por letras y palabras.
    """
    resultado_ordenado = {}

    for letra in sorted(diccionario.keys()):
        palabras = diccionario[letra]

        palabras_ordenadas = dict(sorted(palabras.items()))

        resultado_ordenado[letra] = palabras_ordenadas

    return resultado_ordenado
```

## Función principal

Esta función coordina todo el proceso:

1. Lee el archivo
2. Limpia el texto
3. Cuenta palabras
4. Agrupa resultados
5. Ordena los datos
6. Muestra el resultado

```python
def main():
    """
    Función principal del programa.
    """
    texto = leer_archivo()
    texto = limpiar_texto(texto)

    conteo = contar_palabras(texto)
    agrupado = agrupar_por_letra(conteo)
    resultado = ordenar_resultado(agrupado)

    print("Resultado:\n")

    for letra, palabras in resultado.items():
        print(letra.upper())
        for palabra, cantidad in palabras.items():
            print("  ", palabra, ":", cantidad)
        print()
```

```python
main()
```

## Conclusiones

Este ejercicio muestra un flujo completo de procesamiento de texto:

- Lectura de datos
- Limpieza
- Transformación
- Análisis
- Presentación

El uso de diccionarios permite estructurar la información de forma eficiente, y la separación en funciones facilita la organización del código.

Este tipo de análisis es fundamental en tareas de ciencia de datos y procesamiento de textos.

# Conversión de tipos de datos

## Introducción

En este ejercicio se desarrolla un programa en Python que analiza una lista de elementos y determina cuáles pueden convertirse a valores numéricos.

El programa:

- Recorre una lista que puede contener distintos tipos de datos
- Intenta convertir cada elemento a número (float)
- Separa los valores en dos listas:
  - Convertibles
  - No convertibles

Además, el programa es capaz de procesar estructuras anidadas como listas y tuplas.

Este tipo de procesamiento es habitual en limpieza de datos.

```python
# Imports
```

## Procesamiento de datos

Esta función recorre una lista y clasifica los elementos según si pueden convertirse a número o no.

También gestiona listas y tuplas dentro de la lista principal.

```python
def convertir_datos(lista):
    """
    Convierte los elementos de una lista a float cuando es posible
    y separa los que no se pueden convertir.
    """

    convertibles = []
    no_convertibles = []

    for elemento in lista:

        # si es lista o tupla, recorrer dentro
        if isinstance(elemento, (list, tuple)):

            for sub in elemento:
                try:
                    numero = float(sub)
                    convertibles.append(numero)
                except:
                    no_convertibles.append(sub)

        else:
            try:
                numero = float(elemento)
                convertibles.append(numero)
            except ValueError:
                no_convertibles.append(elemento)

    return convertibles, no_convertibles
```

## Función principal

Esta función define una lista de datos de ejemplo, ejecuta la conversión y muestra los resultados.

```python
def main():
    """
    Función principal del programa.
    """

    datos = ['1.3', 'one', '1e10', 'seven', '3-1/2',
             ('2', 1, 1.4, 'not-a-number'),
             [1, 2, '3', '3.4']]

    buenos, malos = convertir_datos(datos)

    print("Convertibles:", buenos)
    print("No convertibles:", malos)
```

## Ejecución del programa

```python
main()
```

## Conclusiones

Este ejercicio muestra cómo validar y transformar datos en Python.

El uso de try-except permite manejar errores de conversión de forma controlada.

Además, la detección de estructuras anidadas (listas y tuplas) permite procesar datos más complejos, lo cual es muy común en tareas de limpieza y preparación de datos en análisis de datos.

# Generador de contraseñas seguras

## Introducción

En este ejercicio se desarrolla un programa en Python que genera contraseñas aleatorias seguras.

La contraseña se construye en función de varios parámetros:

- Longitud deseada
- Inclusión de mayúsculas
- Inclusión de minúsculas
- Inclusión de números
- Inclusión de caracteres especiales

El programa garantiza que se cumplan los criterios seleccionados y que la contraseña sea aleatoria.

Además, se incluye una funcionalidad extra que permite copiar la contraseña generada al portapapeles.

```python
# Imports
import numpy as np
import pyperclip
```

## Generación de la contraseña

Esta función crea una contraseña aleatoria según los parámetros indicados.

Se asegura de incluir al menos un carácter de cada tipo seleccionado.

```python
def crear_contrasenya(longitud, mayusculas=True, minusculas=True, numeros=True, signos=False):
    """
    Genera una contraseña aleatoria según los parámetros indicados.
    """

    # Listas de caracteres
    mayus = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")
    minus = list("abcdefghijklmnopqrstuvwxyz")
    nums = list("0123456789")
    simb = list("!@#$%&*")

    # Lista final de caracteres disponibles
    caracteres = []

    # Lista para asegurar que hay al menos uno de cada tipo
    contrasenya = []

    # Añadir tipos según parámetros, garantizamos por lo menos un 1 elegido aleatoriamente de cada tipo
    if mayusculas:
        caracteres += mayus
        contrasenya.append(np.random.choice(mayus))

    if minusculas:
        caracteres += minus
        contrasenya.append(np.random.choice(minus))

    if numeros:
        caracteres += nums
        contrasenya.append(np.random.choice(nums))

    if signos:
        caracteres += simb
        contrasenya.append(np.random.choice(simb))

    # Comprobar que hay al menos un tipo seleccionado
    if len(caracteres) == 0:
        return "Error: no has seleccionado ningún tipo de carácter"

    # Rellenar el resto de la contraseña, añade carácteres aleatorios de todos los tipos
    while len(contrasenya) < longitud:
        contrasenya.append(np.random.choice(caracteres))

    # Mezclar la contraseña para evitar patrones, evita que queden en orden
    np.random.shuffle(contrasenya)

    # Convertir lista a string
    return "".join(contrasenya)
```

## Función principal

Esta función genera una contraseña con unos parámetros definidos, la muestra por pantalla y la copia al portapapeles.

```python
def main():
    """
    Función principal del programa.
    """
    password = crear_contrasenya(10, True, True, True, True)

    print("Contraseña generada:", password)

    pyperclip.copy(password)
    print("Contraseña copiada al portapapeles")
```

## Ejecución del programa

```python
main()
```

## Conclusiones

Este ejercicio muestra cómo generar datos aleatorios en Python utilizando la librería NumPy.

La construcción de la contraseña garantiza el cumplimiento de los requisitos definidos, lo que mejora la seguridad.

Además, la integración con el portapapeles añade una funcionalidad práctica para el usuario.

Este tipo de generación de datos es habitual en aplicaciones relacionadas con seguridad y autenticación.

En este ejercicio se desarrolla un programa en Python que procesa un archivo con resultados de partidos de fútbol.

El objetivo es:

- Calcular los goles totales a favor de cada equipo
- Identificar el equipo más goleador

- Generar una clasificación global en función de los puntos

Cada victoria suma 3 puntos, cada empate 1 punto y cada derrota 0 puntos.

El programa está estructurado en funciones para separar la lectura, el procesamiento y el análisis de los datos.

```python
# Imports

from pathlib import Path
```

## Lectura de datos

Esta función lee el archivo de partidos y devuelve todas las líneas.

```python

def leer_archivo():
    ruta = Path.cwd().parent / "historic partits.txt"
    
    if not ruta.exists():
        print("Ruta usada:", ruta)
        raise FileNotFoundError("Archivo no encontrado")
    
    return ruta.read_text(encoding="utf-8").splitlines()
```

- Puntos acumulados

Se utilizan diccionarios para almacenar y acumular los resultados.

```python
def procesar_partidos(lineas):
    """
    Procesa los partidos y calcula goles y puntos por equipo.
    """

    goles_favor = {}
    goles_contra = {}
    puntos = {}

    for linea in lineas:

        equipo1, resultado, equipo2 = linea.strip().split("\t")

        g1, g2 = resultado.split("-")
        g1 = int(g1)
        g2 = int(g2)

        # Inicializar equipos
        for eq in [equipo1, equipo2]:
            if eq not in goles_favor:
                goles_favor[eq] = 0
                goles_contra[eq] = 0
                puntos[eq] = 0

        # Goles
        goles_favor[equipo1] += g1
        goles_contra[equipo1] += g2

        goles_favor[equipo2] += g2
        goles_contra[equipo2] += g1

        # Puntos
        if g1 > g2:
            puntos[equipo1] += 3
        elif g2 > g1:
            puntos[equipo2] += 3
        else:
            puntos[equipo1] += 1
            puntos[equipo2] += 1

    return goles_favor, goles_contra, puntos
```

## Ordenación de resultados

Estas funciones permiten ordenar los resultados por goles y puntos.

```python
def obtener_goles(par):
    """
    Devuelve los goles de un equipo para ordenar. 
    goles_favor.items() da un par ("Granollers", 15)    par[0] = "Granollers", par[1] = 15
    """
    return par[1]


def obtener_puntos(par):
    """
    Devuelve los puntos de un equipo para ordenar.
    """
    return par[1]
```

## Cálculo de resultados

Esta función obtiene:

- El equipo más goleador
- El equipo más goleado
- Los equipos ordenados por goles
- La clasificación ordenada por puntos

```python
def calcular_resultados(goles_favor, goles_contra, puntos):
    """
    Calcula los resultados finales del análisis.
    """

    mas_goleador = max(goles_favor, key=goles_favor.get)
    mas_goleado = max(goles_contra, key=goles_contra.get)

    goles_ordenados = sorted(goles_favor.items(), key=obtener_goles, reverse=True)
    clasificacion = sorted(puntos.items(), key=obtener_puntos, reverse=True)

    return mas_goleador, mas_goleado, goles_ordenados, clasificacion
```

## Función principal

Esta función coordina todo el flujo del programa:

1. Lee el archivo
2. Procesa los datos
3. Calcula resultados
4. Muestra la información

```python
def main():
    """
    Función principal del programa.
    """

    lineas = leer_archivo()

    goles_favor, goles_contra, puntos = procesar_partidos(lineas)

    mas_goleador, mas_goleado, goles_ordenados, clasificacion = calcular_resultados(
        goles_favor, goles_contra, puntos
    )

    print("=== GOLES POR EQUIPO (ordenados) ===")
    for equipo, goles in goles_ordenados:
        print(equipo, ":", goles)

    print("\nEquipo más goleador:", mas_goleador)
    print("Equipo más goleado:", mas_goleado)

    print("\n=== CLASIFICACIÓN GLOBAL ===")
    for equipo, pts in clasificacion:
        print(equipo, ":", pts)
```
