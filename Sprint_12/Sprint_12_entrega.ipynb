# EJERCICIO 1 Peticions GET

```python
# Instal·lació de requests (si cal)
# !pip install requests

import requests

BASE_URL = "https://jsonplaceholder.typicode.com"
```

## Obtenir publicacions (/posts)

```python
posts_response = requests.get(f"{BASE_URL}/posts")

posts = posts_response.json()

print("Codi d'estat POSTS:", posts_response.status_code)
print("Quantitat de publicacions:", len(posts))
```

## Obtenir usuaris (/users)

```python
users_response = requests.get(f"{BASE_URL}/users")

users = users_response.json()

print("Codi d'estat USERS:", users_response.status_code)
print("Quantitat d'usuaris:", len(users))
```

## Obtenir tasques (/todos)

```python
todos_response = requests.get(f"{BASE_URL}/todos")

todos = todos_response.json()

print("Codi d'estat TODOS:", todos_response.status_code)
print("Quantitat de tasques:", len(todos))
```

# EJERCCIO 2. Mostrar quantitats totals i codis d'estat

```python
print("===== RESUM =====")

print(f"POSTS -> Total: {len(posts)} | Status: {posts_response.status_code}")

print(f"USERS -> Total: {len(users)} | Status: {users_response.status_code}")

print(f"TODOS -> Total: {len(todos)} | Status: {todos_response.status_code}")
```

# EJERCICIO 3. Petició a una publicació inexistent (error 404)

```python
error_response = requests.get(f"{BASE_URL}/posts/999999")

print("Codi d'estat error:", error_response.status_code)
```

# EJERCICIO 4. Petició POST per crear una nova publicació

```python
new_post = {
    "title": "La meva nova publicació",
    "body": "Aquest és el contingut de la publicació.",
    "userId": 1
}

post_response = requests.post(f"{BASE_URL}/posts", json=new_post)

print("Codi d'estat POST:", post_response.status_code)

print("Resposta JSON:")
print(post_response.json())
```

# EJERCICIO 5. Petició PATCH per modificar parcialment una publicació

```python
update_data = {
    "title": "Títol modificat"
}

patch_response = requests.patch(
    f"{BASE_URL}/posts/1",
    json=update_data
)

print("Codi d'estat PATCH:", patch_response.status_code)

print("Resposta JSON:")
print(patch_response.json())
```

# EJERCICIO 3. Petició DELETE sobre una publicació

```python
delete_response = requests.delete(f"{BASE_URL}/posts/1")

print("Codi d'estat DELETE:", delete_response.status_code)

print("Resposta JSON:")
print(delete_response.json())
```

# EJERCICIO 1

https://restcountries.com/

Esta API permite obtener información sobre países mediante peticiones GET.

# EJERCICIO 2. Documentación de la API

## Endpoints disponibles

GET /v3.1/all
→ Devuelve información de todos los países.

GET /v3.1/name/{name}
→ Busca países por nombre.

## Parámetros y filtros interesantes

fields=
→ Permite seleccionar únicamente algunos campos.

Ejemplo:
https://restcountries.com/v3.1/all?fields=name,capital,population

region=
→ Permite filtrar por región.

Ejemplo:
https://restcountries.com/v3.1/region/europe

## Formato de respuesta

La API devuelve las respuestas en formato JSON.

# EJERCICIO 3. Petición GET sencilla

## Importar librerías

```python
import requests
import pandas as pd
```

## Hacer petición GET

```python
url = "https://restcountries.com/v3.1/all?fields=name,capital,population,region"

response = requests.get(url)

print("Código de estado:", response.status_code)
```

## Convertir respuesta JSON

```python
countries = response.json()

print("Cantidad de países:", len(countries))
```

## Mostrar algunos campos del JSON

```python
for country in countries[:5]:

    print("País:", country["name"]["common"])

    capital = country.get("capital", ["Sin capital registrada"])[0]

    print("Capital:", capital)

    print("Población:", country.get("population"))

    print("-" * 40)
```

# EJERCICIO 4. Convertir datos a DataFrame de pandas

## Crear lista con datos simplificados

```python
import pandas as pd

data = []

for country in countries:

    capitales = country.get("capital", [])

    capital = capitales[0] if len(capitales) > 0 else "Sin capital"

    data.append({
        "Pais": country["name"]["common"],
        "Capital": capital,
        "Poblacion": country.get("population"),
        "Region": country.get("region")
    })
```

## Crear DataFrame

```python
df = pd.DataFrame(data)
```

## Mostrar primeras filas

```python
df.head()
```

# EJERCICIO 1. Buscar un dataset con package_search

## Importar librerías

```python
import requests
import pandas as pd
```

## URL base de la API

```python
BASE_URL = "https://opendata-ajuntament.barcelona.cat/data/api/3/action"
```

## Buscar datasets relacionados con "bicing"

```python
search_url = f"{BASE_URL}/package_search"

params = {
    "q": "bicing"
}

response = requests.get(search_url, params=params)

print("Código de estado:", response.status_code)
```

## Convertir respuesta a JSON

```python
data = response.json()
```

## Mostrar algunos resultados encontrados

```python
results = data["result"]["results"]

print("Cantidad de datasets encontrados:", len(results))

for dataset in results[:3]:
    print("Título:", dataset["title"])
    print("ID:", dataset["id"])
    print("-" * 50)
```

# EJERCICIO 2. Obtener detalles del dataset con package_show

## Seleccionamos el primer dataset encontrado.

```python
dataset_id = results[0]["id"]

show_url = f"{BASE_URL}/package_show"

params = {
    "id": dataset_id
}

response_show = requests.get(show_url, params=params)

print("Código de estado:", response_show.status_code)
```

## Ver recursos disponibles

```python
dataset_info = response_show.json()

resources = dataset_info["result"]["resources"]

for resource in resources:
    print("Nombre:", resource["name"])
    print("Formato:", resource["format"])
    print("Resource ID:", resource["id"])
    print("-" * 50)
```
